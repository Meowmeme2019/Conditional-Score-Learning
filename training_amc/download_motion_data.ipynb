```python
import os
import requests
from bs4 import BeautifulSoup
import re
```

```python
# Save base path
base_save_path = r"C:\Users\chenw\OneDrive - University of Pittsburgh\_score network code\Markov chain neumeric\CMU motion capture\raw data"
os.makedirs(base_save_path, exist_ok=True)

# Trial ID range to check
trial_range = range(1, 99)  # Try 01 to 50

# Subjects 1 to 144
for subject_num in range(1, 145):
    subject_id = f"{subject_num:02d}"  # zero-padded, e.g., "01"
    subject_folder = os.path.join(base_save_path, f"Subject_{subject_id}")
    os.makedirs(subject_folder, exist_ok=True)

    base_url = f"http://mocap.cs.cmu.edu/subjects/{subject_id}/"

    print(f" Processing Subject {subject_id}")

    # Download ASF (skeleton file)
    asf_url = f"{base_url}{subject_id}.asf"
    asf_path = os.path.join(subject_folder, f"{subject_id}.asf")
    r = requests.get(asf_url)
    if r.status_code == 200:
        with open(asf_path, 'wb') as f:
            f.write(r.content)
        # print(f" ASF downloaded for Subject {subject_id}")
    else:
        print(f"ASF file not found for Subject {subject_id} (URL: {asf_url})")
        continue  # skip subject if no skeleton

    # Try downloading AMC trials
    for tid in trial_range:
        trial_id = f"{tid:02d}"
        amc_url = f"{base_url}{subject_id}_{trial_id}.amc"
        amc_filename = f"Trial_{trial_id}.amc"
        amc_path = os.path.join(subject_folder, amc_filename)

        r = requests.get(amc_url)
        if r.status_code == 200:
            with open(amc_path, 'wb') as f:
                f.write(r.content)
            # print(f"   {amc_filename} downloaded.")
        else:
            # Uncomment below if you want to print missing files
            # print(f"  Trial {trial_id} not found.")
            pass

print(" Done downloading all available data from CMU MoCap!")
```
